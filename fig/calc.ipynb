```julia
f(x) = real(SF.gamma(x))
g(x) = imag(SF.gamma(x))

p = plot(ylims=(-10, 10), xlims=(-5.1, 5), xticks=-5:1:5,
         xlabel=L"x", ylabel=L"\Gamma(x)", legend=:topright)

# 分段绘制，避开 x = -5, -4, -3, -2, -1, 0 的奇点
segments = [(-5.5, -5.05), (-4.999999, -4.000000001), (-3.999995, -3.0000001), 
            (-2.9999999995, -2.000005), (-1.95, -1.05), (-0.95, 0.05), (0.1, 5)]

for (i, seg) in enumerate(segments)
    xs_seg = seg[1]:0.01:seg[2]
    ys_seg = f.(xs_seg)
    y2s_seg = g.(xs_seg)
    
    plot!(p, xs_seg, ys_seg, label=(i==1 ? L"Re(\Gamma(x))" : ""), linewidth=2)

end

display(p)
```
