```julia
using Plots
using LaTeXStrings
```

```julia
X1s = 1:0.1:3
Y1s = 1:0.1:3


function f(x, y)
    term1 = x*y 
    term2 = sqrt(x^2-1) * sqrt(y^2-1)
    return term1 - term2
end

Zs = [f(x, y) for x in X1s, y in Y1s]

heatmap(X1s, Y1s, Zs, c=:blues, xlabel=L"\Lambda_1", ylabel=L"\Lambda_2")
savefig("Lambda00.svg")
```
